## DISPLAY AN IMAGE

```python
image=cv.imread("tank.png")
cv.imshow("My Image",image)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
blue = image[100, 200, 0]
green = image[100, 200, 1]
red = image[100, 200, 2]

print("Blue:", blue)
print("Green:", green)
print("Red:", red)
```

### CROP AN IMAGE

```python
crop=image[100:400, 200:600]
cv.imwrite("crop.jpg", crop)

cv.imshow("Crop", crop)

cv.waitKey(0)
cv.destroyAllWindows()
```

# READING A VIDEO

```python
vidpath = "model_train.mp4" 
cap = cv.VideoCapture(vidpath) 

while cap.isOpened(): 
    ret, frame = cap.read() 
    
    # Break the loop if the video ends or fails to load
    if not ret:
        break
        
    cv.imshow("video", frame) 
    
    delay = int(1000 / 60) 
    if cv.waitKey(delay) == ord('q'): 
        break

cap.release()
cv.destroyAllWindows()
```

# SHAPES AND TEXT

```python
image=cv.imread("tank.png")

#!             start ,   end   ,  color  ,thickness
cv.line(image,(50,50),(200,300),(0,255,0),3)

cv.rectangle(image,(50,50),(200,300),(0,255,0),3)

cv.circle(image,(50,50),80,(0,255,0),3)

cv.ellipse(image,(300, 200),(100, 50),0,0,360,(0, 255, 255),3)
cv.imshow("line",image)
cv.waitKey(0)
cv.destroyAllWindows()
```

#### DRAW TEXT

```python
image=cv.imread("tank.png")
#!                 text     , position ,  font  , font_scale , color , thickness
cv.putText(image,"HIIIII!!!",(50,100),cv.FONT_HERSHEY_COMPLEX,1,(0,255,0),2)
cv.imshow("line",image)
cv.waitKey(0)
cv.destroyAllWindows()
```

# BGR, RGB, Grayscale & HSV

```python
image=cv.imread("tank.png")
rgb=cv.cvtColor(image,cv.COLOR_BGR2RGB)

cv.imshow("Gray", rgb)
cv.waitKey(0)
cv.destroyAllWindows()


print("OG",image.shape)
print("RGB",rgb.shape)
```

```python
gray=cv.cvtColor(image,cv.COLOR_BGR2GRAY)
cv.imshow("Gray", gray)
cv.waitKey(0)
cv.destroyAllWindows()

print("OG",image.shape)
print("RGB",gray.shape)
```

```python
hsv=cv.cvtColor(image,cv.COLOR_BGR2HSV)
cv.imshow("Gray", hsv)
cv.waitKey(0)
cv.destroyAllWindows()

print("OG",image.shape)
print("RGB",hsv.shape)
```
