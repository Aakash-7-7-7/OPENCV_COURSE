### BRG2GRAY CONVERTER

```python
def convert_bgr2gray(path):
    path=cv.imread(path)
    inp=int(input("ENTER A number , 1: for gray , 2: for HSV"))

    if inp==1:
        gray=cv.cvtColor(path , cv.COLOR_BGR2GRAY)
        gray=cv.resize(gray , None , fx=0.3 , fy=0.3 , interpolation=cv.INTER_AREA)
        cv.imshow("gray image",gray)
        

    elif inp==2:
        hsv=cv.cvtColor(path , cv.COLOR_BGR2HSV)
        hsv=cv.resize(hsv , None , fx=0.3 , fy=0.3 , interpolation=cv.INTER_AREA)
        cv.imshow("HSV image",hsv)

    elif inp==3:
        gray=cv.cvtColor(path , cv.COLOR_BGR2GRAY)
        gray=cv.resize(gray , None , fx=0.3 , fy=0.3 , interpolation=cv.INTER_AREA)

        adaptive =cv.adaptiveThreshold(gray , 255 , cv.ADAPTIVE_THRESH_MEAN_C ,cv.THRESH_BINARY , 5 ,3)
        cv.imshow("Threshold", adaptive)
    
    else:
        print("ENTER CORRECT NUMBER")


    cv.waitKey(0)                
    cv.destroyAllWindows()


convert_bgr2gray("IMAGES/mem.png")
```

## THRESHOLDING

## MORPHOLOGICAL OPERATION

```python
image=cv.resize(cv.imread("IMAGES/dilation.png"), None , fx=4 , fy=4 , interpolation=cv.INTER_AREA)
gray=cv.cvtColor(image , cv.COLOR_BGR2GRAY)

_,binary=cv.threshold(gray , 127,255 , cv.THRESH_BINARY)

kernel=cv.getStructuringElement(cv.MORPH_OPEN , (5,5))
eroded=cv.erode(binary , kernel , iterations=3)

cv.imshow("og",image)
cv.imshow("imahe",eroded)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
def fix_text(path):
    
    image=cv.resize(cv.imread(path), None , fx=0.3 , fy=0.3 , interpolation=cv.INTER_AREA)

    gray=cv.cvtColor(image , cv.COLOR_BGR2GRAY)
    _,binary=cv.threshold(gray , 127 , 255  , cv.THRESH_BINARY)

    kernel=cv.getStructuringElement(cv.MORPH_OPEN,(4,4))
    

    opening=cv.morphologyEx(binary , cv.MORPH_OPEN , kernel)
    closing=cv.morphologyEx(opening , cv.MORPH_CLOSE ,kernel)

    cv.imshow("final", closing)
    cv.waitKey(0)
    cv.destroyAllWindows()



fix_text("IMAGES/long.png")
```

```python
image=cv.resize(cv.imread("IMAGES/coins.png") , None , fx=0.6 , fy=0.6 , interpolation=cv.INTER_AREA)

gray=cv.cvtColor(image , cv.COLOR_BGR2GRAY)
blur=cv.bilateralFilter(gray ,9,75,75)

edges=cv.Canny(blur , 50,200)
kernel=cv.getStructuringElement(cv.MORPH_RECT , (3,3))
closed_edges=cv.morphologyEx(edges , cv.MORPH_CLOSE , kernel)


contours, heirarchy=cv.findContours(edges , cv.RETR_EXTERNAL , cv.CHAIN_APPROX_SIMPLE)

cv.drawContours(image , contours , -1,(0,255,0),2)
cv.imshow("edges",edges)
cv.imshow("Detected Contours",image)

cv.waitKey(0)
cv.destroyAllWindows()
```
